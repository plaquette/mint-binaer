## 2 - Geheimwort decodieren

Diese vier Bytes ergeben ein Wort. **Rate zuerst** mit der Tabelle, dann prüfe.

```python
bytes_liste = ['01000010', '01001001', '01010100', '01010011']

wort = ''
for b in bytes_liste:
    wort = wort + chr(int(b, 2))
print('Das Wort heisst:', wort)
```

## 3 - Wie gross ist ein Text?

Jedes Zeichen (auch das Leerzeichen!) braucht 1 Byte.

```python
satz = 'MINT macht Spass'

print('Zeichen:', len(satz))
print('Byte:', len(satz))
print('Bit:', len(satz) * 8)
print('In 1 KB (1024 Byte) passen', 1024 // len(satz), 'solche Sätze.')
```

## 4 - Ein Bit umklappen

`^` klappt ein Bit um. Beim Stellenwert **32** passiert etwas Praktisches.

```python
zeichen = 'A'
print(zeichen, '=', format(ord(zeichen), '08b'))

neu = chr(ord(zeichen) ^ 32)
print(neu, '=', format(ord(neu), '08b'))
```

```python
# Ein ganzes Wort umschalten - probiere auch ein Wort in Kleinbuchstaben:
wort = 'COMPUTER'
print(''.join(chr(ord(z) ^ 32) for z in wort))
```

## R2 - Dein eigenes Bild
Ein Bild ist eine Liste von Helligkeiten (0 = schwarz, 255 = weiss).
Baue ein 8 x 8 Muster: ändere die Zahlen und schau, was entsteht.

```python
# --- Technik-Zelle: einmal ausführen, nicht erklären -----------------------
# Baut aus einer Liste von Zahlen (0-255) ein Graustufenbild. Nur Standardbibliothek.
import zlib, struct
from IPython.display import Image, display

def png_block(typ, data):
    return struct.pack('>I', len(data)) + typ + data + struct.pack('>I', zlib.crc32(typ + data))

def zeige_bild(pixel, breite, hoehe, zoom=4):
    zeilen = b''
    for y in range(hoehe):
        reihe = b''.join(bytes([p]) * zoom for p in pixel[y*breite:(y+1)*breite])
        zeilen += (b'\x00' + reihe) * zoom
    kopf = struct.pack('>IIBBBBB', breite*zoom, hoehe*zoom, 8, 0, 0, 0, 0)
    png = (b'\x89PNG\r\n\x1a\n' + png_block(b'IHDR', kopf)
           + png_block(b'IDAT', zlib.compress(zeilen)) + png_block(b'IEND', b''))
    display(Image(data=png))

print('Bereit.')
```

```python
muster = [
    0,   0, 255, 255, 255, 255,   0,   0,
    0, 255, 255, 255, 255, 255, 255,   0,
  255, 255,   0, 255, 255,   0, 255, 255,
  255, 255, 255, 255, 255, 255, 255, 255,
  255,   0, 255, 255, 255, 255,   0, 255,
  255, 255,   0,   0,   0,   0, 255, 255,
    0, 255, 255, 255, 255, 255, 255,   0,
    0,   0, 255, 255, 255, 255,   0,   0,
]
zeige_bild(muster, 8, 8, zoom=20)
print('Dieses Bild braucht', len(muster), 'Byte.')
```
