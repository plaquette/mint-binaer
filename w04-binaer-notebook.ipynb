### A1 - Stellenwerte: so liest der Computer eine Binärzahl
Im Zweiersystem verdoppelt sich der Stellenwert von rechts nach links: **1, 2, 4, 8, 16, 32 ...**
Nur die Stellen mit einer **1** zählen mit.

```python
# Stellenwerte für 6 Stellen (wie die 6 Karten)
stellenwerte = [32, 16, 8, 4, 2, 1]
print(stellenwerte)
```

```python
# Binärzahl 010010  -> welche Stellen sind 1?
binaer = '010010'
summe = 0
for ziffer, wert in zip(binaer, stellenwerte):
    if ziffer == '1':
        print(ziffer, 'an der Stelle', wert, '  ->  zählt', wert)
        summe += wert
    else:
        print(ziffer, 'an der Stelle', wert, '  ->  zählt 0')
print('---')
print('Zusammen:', summe)
```

**Vorhersage vorne:** Bevor wir ausführen - was gibt `010010` im Zehnersystem?  
(Auflösung: die Zelle ausführen.)

### A2 - Python rechnet selbst um
Python kann direkt umrechnen. Wir vergleichen mit unserem Ergebnis von Hand.

```python
# Zweiersystem -> Zehnersystem
print('010010 im Zweiersystem =', int('010010', 2), 'im Zehnersystem')

# Zehnersystem -> Zweiersystem (bin() hängt '0b' davor, das schneiden wir ab)
print('18 im Zehnersystem =', bin(18)[2:], 'im Zweiersystem')
```

### A3 - Das Subtraktionsverfahren (Zehner -> Zweier)
So habt ihr es in Woche 3/4 von Hand gemacht: immer die **grösste passende Zweierpotenz abziehen**.
Python macht jeden Schritt sichtbar.

```python
def subtraktionsverfahren(zahl):
    rest = zahl
    ziffern = ''
    for wert in [32, 16, 8, 4, 2, 1]:
        if rest >= wert:
            ziffern += '1'
            print(rest, '-', wert, '=', rest - wert, '  -> Stelle', wert, 'bekommt eine 1')
            rest -= wert
        else:
            ziffern += '0'
            print(rest, 'ist kleiner als', wert, '        -> Stelle', wert, 'bekommt eine 0')
    print('Ergebnis:', ziffern)
    return ziffern

subtraktionsverfahren(53)
```

### A4 - Und jetzt der Kartentrick
Beim Trick gibt es **6 Karten**. In der Ecke jeder Karte steht eine Zweierpotenz: **1, 2, 4, 8, 16, 32**.
Auf einer Karte stehen genau die Zahlen, die an dieser Stelle eine **1** haben.

Die nächste Zelle zeigt, **auf welchen Karten eine Zahl steht**.

```python
def karten_fuer(zahl):
    # gibt die Eckzahlen der Karten zurück, auf denen 'zahl' steht
    ecken = []
    for wert in [32, 16, 8, 4, 2, 1]:
        if zahl % (2 * wert) >= wert:   # steht an dieser Binärstelle eine 1?
            ecken.append(wert)
    return ecken

zahl = 7
ecken = karten_fuer(zahl)
print('Die', zahl, 'steht auf den Karten mit Eckzahl:', ecken)
print('Eckzahlen addiert:', sum(ecken))
```

**Das ist der ganze Trick:** Die Person gibt euch die Karten, auf denen ihre Zahl steht.
Ihr addiert nur die **Eckzahlen** - und habt die Zahl. Warum? Weil die Eckzahlen genau die
Binärstellen sind, die eine 1 haben. **Die Karten lesen die Binärdarstellung der Zahl ab.**

### R1 - Der Trick mit 7 Karten (Zahlen bis 127)
Mit 6 Karten reichen wir bis 63. Für Zahlen bis 127 braucht es eine **7. Karte**.
Welche Eckzahl steht darauf? (Tipp: die nächste Zweierpotenz.)
Vervollständige die Liste `stellen7` und führe aus - die Selbstkontrolle sagt dir, ob es stimmt.

```python
# Ergänze die fehlende Eckzahl (statt der None):
stellen7 = [None, 32, 16, 8, 4, 2, 1]

def karten_fuer_7(zahl):
    ecken = []
    for wert in stellen7:
        if wert is not None and zahl % (2 * wert) >= wert:
            ecken.append(wert)
    return ecken

print('100 steht auf den Karten:', karten_fuer_7(100), ' Summe:', sum(karten_fuer_7(100)))
assert stellen7[0] == 64, 'Welche Zweierpotenz kommt nach 32? (2 mal 32)'
assert sum(karten_fuer_7(100)) == 100, 'Noch nicht - prüfe die Eckzahl'
print('Stimmt! Mit 7 Karten (Eckzahl 64 dazu) klappt der Trick bis 127.')
```

### R3 - Warum ausgerechnet 1, 2, 4, 8, 16, 32?
Das ist die eigentliche Warum-Frage. Was ginge schief, wenn wir als Eckzahlen z. B.
**1, 2, 3, 4, 5, 6** nähmen statt der Zweierpotenzen?

Die Zelle sucht eine Zahl, die sich mit 1, 2, 3, 4, 5, 6 auf **zwei verschiedene Arten** legen lässt.
Bei den Zweierpotenzen gibt es das **nie** - jede Zahl hat genau eine Darstellung. Das ist der Kern des Tricks.

```python
from itertools import combinations

def anzahl_darstellungen(zahl, ecken):
    # auf wie viele Arten lässt sich 'zahl' als Summe verschiedener Eckzahlen schreiben?
    treffer = 0
    for r in range(1, len(ecken) + 1):
        for auswahl in combinations(ecken, r):
            if sum(auswahl) == zahl:
                treffer += 1
    return treffer

schlecht = [1, 2, 3, 4, 5, 6]
gut = [1, 2, 4, 8, 16, 32]

print('Mit schlechten Eckzahlen', schlecht, '- Zahlen mit MEHR als einer Darstellung:')
print([z for z in range(1, 22) if anzahl_darstellungen(z, schlecht) > 1][:8], '...')
print()
print('Mit Zweierpotenzen', gut, '- hat irgendeine Zahl mehr als eine Darstellung?')
print('Maximum:', max(anzahl_darstellungen(z, gut) for z in range(1, 64)))
print('-> immer genau 1. Deshalb ist die Zahl aus den Karten eindeutig ablesbar.')
```

**Das ist der tiefe Grund:** Nur Zweierpotenzen geben jeder Zahl **genau eine** Darstellung -
das ist das Binärsystem. Andere Eckzahlen wären mehrdeutig, und der Trick würde manchmal versagen.
